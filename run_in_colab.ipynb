Try it: change the doping, oxide thickness or body thickness and re-run the cells below.

```python
from mosdev import MOSFET, extract
import numpy as np
vg = np.linspace(0, 1, 2001)
for tox in [1.0, 1.5, 2.5]:
    m = MOSFET(1.0, 60.0, tox_nm=tox)
    print(f'tox {tox} nm  DIBL {extract.dibl(m.vth(0.05), m.vth(1.0), 0.05, 1.0):6.1f} mV/V  SS {extract.subthreshold_swing(vg, m.id(vg, 1.0), 1e-16):5.1f} mV/dec')
```
